```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("dongeorge/beer-consumption-sao-paulo")

print("Path to dataset files:", path)
```

```python
import os
import pandas as pd

# Construct the full path to the CSV file
file_path = os.path.join(path, 'Consumo_cerveja.csv')

# Load the CSV file into a pandas DataFrame
df = pd.read_csv(file_path)

# Display the first 5 rows of the DataFrame
display(df.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a scatter plot with a linear regression line
plt.figure(figsize=(10, 6))
sns.regplot(x='avg_temp_c', y='beer_consumption_l', data=df, scatter_kws={'alpha':0.6}, line_kws={'color':'red'})
plt.title('Linear Regression: Beer Consumption vs. Average Temperature')
plt.xlabel('Average Temperature (C)')
plt.ylabel('Beer Consumption (liters)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```
